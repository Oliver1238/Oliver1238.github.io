**Ejercicio 6**: Escribir un método en la clase Mazo que *mezcle* el mazo. Puede ser de utilidad el módulo [`random`](https://docs.python.org/3/library/random.html) de la biblioteca estándar de Python.

```python
import random  # Importamos el módulo necesario para mezclar

class Carta:
    def __init__(self, valor, descripcion_tipo, palo):
        self.valor = valor
        self.descripcion_tipo = descripcion_tipo
        self.palo = palo
        self.numero = valor
        self.apodos = {
            1: "As", 11: "Caballo", 12: "Rey", 10: "Sota",
            2: "Dos", 3: "Tres", 4: "Cuatro", 5: "Cinco",
            6: "Seis", 7: "Siete",
        }

    def __str__(self):
        valor_display = self.apodos.get(self.valor, str(self.valor))
        return f"{valor_display} de {self.palo}"


    def __eq__(self, otra_carta):
        if not isinstance(otra_carta, Carta):
            return False


        return self.valor == otra_carta.valor and self.palo == otra_carta.palo

ancho_de_espadas = Carta(1, "Carta-alta", "Espada")
ancho_de_basto = Carta(1, "Carta alta", "Basto")
otro_ancho_de_espadas = Carta(1, "Otra copia", "Espada")
as_de_espadas = Carta(1, "Carta alta", "Espada")

print(ancho_de_espadas == ancho_de_basto)
print(ancho_de_espadas == otro_ancho_de_espadas)
print(ancho_de_espadas == "Un string cualquiera")
print(ancho_de_espadas == as_de_espadas)


class Mazo:
  def __init__(self):
    self.cartas = []
    palos = ["Espada", "Basto", "Oro", "Copa"]
    numeros_validos = [1, 2, 3, 4, 5, 6, 7, 10, 11, 12]

    for palo in palos:
        for numero in numeros_validos:
            self.cartas.append(Carta(numero, "Común", palo))

  def cantidad (self):
    return len(self.cartas)

  def mezclar(self):
    random.shuffle(self.cartas)
    return self.cartas

  def sacar_carta(self):
    if self.cantidad() > 0:
        return self.cartas.pop()
    else:
        return None


mi_mazo = Mazo()
print(f"Cantidad de cartas en el mazo antes de mezclar: {mi_mazo.cantidad()}")
print(f"Primera carta antes de mezclar: {mi_mazo.cartas[0]}")
for i, carta in enumerate(mi_mazo.cartas):
    print(f"Carta {i+1}: {carta}")

mi_mazo.mezclar()
print(f"\nCantidad de cartas en el mazo después de mezclar: {mi_mazo.cantidad()}")

for i, carta in enumerate(mi_mazo.cartas):
    print(f"Carta {i+1}: {carta}")

carta_sacada = mi_mazo.sacar_carta()
print(f"\nSe sacó la carta: {carta_sacada}")
print(f"Cantidad de cartas restantes en el mazo: {mi_mazo.cantidad()}")
```
